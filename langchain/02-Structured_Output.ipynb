## Structured Output

Sometimes we want that Models provide the formatted output or schema based output instead of the raw output. It is useful to improve analytical processesing, avoiding unnecessary output filtration, reduce costing etc.

There are mainly 3 ways to acheive structured output :-
1) Pydantic 
2) TypeDict
3) Data Classes

### Pydantic

Pydantic is the most popular data validation, automatic type coercion and settings management library for Python.


```python
import os
from dotenv import load_dotenv

load_dotenv()
os.environ["GROQ_API_KEY"] = os.getenv("Groq_API_KEY")
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent

llm = ChatGroq(
    model = "openai/gpt-oss-20b",
    temperature= 0,
)
```

```python
response = llm.invoke("Tell me about movie 'Inception'.")
response
```

```python
from pydantic import BaseModel, Field

class Movie(BaseModel):
    name:str = Field(description="The movie name")
    year:str = Field(description="The year movie was released")
    director:str = Field(description="Movie Director's name")
    rating:float = Field(description="Movie's Rating out of 10")
    budget:str = Field(description="Movie's Buget")
    genres: list[str] = Field(description="Movie's related generes")


llm_with_structuredOutput = llm.with_structured_output(Movie)

response = llm_with_structuredOutput.invoke("Tell me about movie 'Inception'.")
response
```

#### Nested Structure

```python
from pydantic import BaseModel, Field

class Actor(BaseModel):
    name: str = Field(description="Real Name of the Actor")
    role: str = Field(description="Movie Name of the Actor")

class Movie(BaseModel):
    name:str = Field(description="The movie name")
    year:str = Field(description="The year movie was released")
    director:str = Field(description="Movie Director's name")
    rating:float = Field(description="Movie's Rating out of 10")
    budget:str = Field(description="Movie's Buget")
    cast: list[Actor]
    genres: list[str] = Field(description="Movie's related generes")


llm_with_structuredOutput = llm.with_structured_output(Movie)

response = llm_with_structuredOutput.invoke("Tell me about movie 'Interstellar'.")
response
```
